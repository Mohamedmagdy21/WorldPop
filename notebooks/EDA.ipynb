```python
import rasterio
import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq

PATH = "/home/mohamed/Downloads/egy_ppp_2020_UNadj_constrained.tif"
OUT = "/tmp/opencode/egy_pop_full.parquet"

writer = None
with rasterio.open(PATH) as src:
    transform = src.transform
    # Use the raster's internal tiling; fall back to 2048x2048 strips
    for ji, window in src.block_windows(1):
        data = src.read(1, window=window)
        h, w = data.shape
        r0, c0 = window.row_off, window.col_off

        col = np.arange(c0, c0 + w, dtype=np.float32)
        row = np.arange(r0, r0 + h, dtype=np.float32)
        cc, rr = np.meshgrid(col, row)

        # Pixel-center coordinates via the affine transform
        xs = (transform.c + (cc + 0.5) * transform.a
              + (rr + 0.5) * transform.b).astype(np.float32, copy=False)
        ys = (transform.f + (cc + 0.5) * transform.d
              + (rr + 0.5) * transform.e).astype(np.float32, copy=False)

        table = pa.table({
            "longitude": pa.array(xs.ravel()),
            "latitude": pa.array(ys.ravel()),
            "population": pa.array(data.ravel().astype(np.float32, copy=False)),
        })

        if writer is None:
            writer = pq.ParquetWriter(OUT, table.schema)
        writer.write_table(table)

writer.close()
print("wrote", OUT)

# Sanity check: read a few rows back without loading everything
print(pq.read_table(OUT).slice(0, 5).to_pandas())
```

```python
con.sql("""
WITH valid AS (
    SELECT population
    FROM '/tmp/opencode/egy_pop_full.parquet'
    WHERE population >= 0
),

ranked AS (
    SELECT
        population,
        SUM(population) OVER () AS total_population,
        NTILE(100) OVER (ORDER BY population DESC) AS percentile_group
    FROM valid
)

SELECT
    CASE
        WHEN percentile_group = 1 THEN 'Top 1%'
        WHEN percentile_group <= 5 THEN 'Top 5%'
        WHEN percentile_group <= 10 THEN 'Top 10%'
    END AS group_name,

    COUNT(*) AS pixels,
    SUM(population) AS population,
    SUM(population) / MAX(total_population) * 100 AS population_share_pct

FROM ranked
WHERE percentile_group <= 10
GROUP BY group_name
ORDER BY
    CASE group_name
        WHEN 'Top 1%' THEN 1
        WHEN 'Top 5%' THEN 2
        WHEN 'Top 10%' THEN 3
    END;
""")
```

```python
df_concentration = con.sql("""
WITH valid AS (
    SELECT population
    FROM '/tmp/opencode/egy_pop_full.parquet'
    WHERE population >= 0
),

ranked AS (
    SELECT
        population,
        SUM(population) OVER () AS total_population,
        NTILE(100) OVER (ORDER BY population DESC) AS percentile_group
    FROM valid
),

results AS (
    SELECT
        'Top 1%' AS group_name,
        1 AS cutoff,
        population,
        total_population,
        percentile_group
    FROM ranked

    UNION ALL

    SELECT
        'Top 5%' AS group_name,
        5 AS cutoff,
        population,
        total_population,
        percentile_group
    FROM ranked

    UNION ALL

    SELECT
        'Top 10%' AS group_name,
        10 AS cutoff,
        population,
        total_population,
        percentile_group
    FROM ranked
)

SELECT
    group_name,
    COUNT(*) AS pixels,
    SUM(population) AS population,
    SUM(population) / MAX(total_population) * 100 AS population_share_pct

FROM results

WHERE percentile_group <= cutoff

GROUP BY
    group_name,
    cutoff

ORDER BY cutoff;
""").df()
```

```python
df_concentration_location = con.sql("""
WITH valid AS (
    SELECT
        longitude,
        latitude,
        population
    FROM '/tmp/opencode/egy_pop_full.parquet'
    WHERE population >= 0 AND population <=5000
),

ranked AS (
    SELECT
        longitude,
        latitude,
        population,
        SUM(population) OVER () AS total_population,
        NTILE(100) OVER (ORDER BY population DESC) AS percentile_group
    FROM valid
),

results AS (
    SELECT
        'Top 1%' AS group_name,
        1 AS cutoff,
        longitude,
        latitude,
        population,
        total_population,
        percentile_group
    FROM ranked

    UNION ALL

    SELECT
        'Top 5%' AS group_name,
        5 AS cutoff,
        longitude,
        latitude,
        population,
        total_population,
        percentile_group
    FROM ranked

    UNION ALL

    SELECT
        'Top 10%' AS group_name,
        10 AS cutoff,
        longitude,
        latitude,
        population,
        total_population,
        percentile_group
    FROM ranked
)

SELECT
    group_name,

    COUNT(*) AS pixels,

    SUM(population) AS population,

    -- Percentage of total population
    SUM(population) / MAX(total_population) * 100
        AS population_share_pct,

    -- Population-weighted longitude
    SUM(longitude * population) / SUM(population)
        AS weighted_longitude,

    -- Population-weighted latitude
    SUM(latitude * population) / SUM(population)
        AS weighted_latitude

FROM results

WHERE percentile_group <= cutoff

GROUP BY
    group_name,
    cutoff

ORDER BY cutoff;
""").df()
```

```python
df_top_pixels = con.sql("""
WITH valid AS (
    SELECT
        longitude,
        latitude,
        population
    FROM '/tmp/opencode/egy_pop_full.parquet'
    WHERE population >= 0
),

ranked AS (
    SELECT
        longitude,
        latitude,
        population,
        NTILE(100) OVER (ORDER BY population DESC) AS percentile_group
    FROM valid
)

SELECT
    CASE
        WHEN percentile_group = 1 THEN 'Top 1%'
        WHEN percentile_group BETWEEN 2 AND 5 THEN 'Top 1–5%'
        WHEN percentile_group BETWEEN 6 AND 10 THEN 'Top 5–10%'
    END AS group_name,

    longitude,
    latitude,
    population

FROM ranked

WHERE percentile_group <= 10

ORDER BY percentile_group;
""").df()
```

```python
lorenz_df = con.sql("""
WITH valid AS (
    SELECT population
    FROM '/tmp/opencode/egy_pop_full.parquet'
    WHERE population >= 0
      AND population <= 5000
),

ranked AS (
    SELECT
        population,

        ROW_NUMBER() OVER (
            ORDER BY population
        ) AS pixel_rank,

        COUNT(*) OVER () AS total_pixels,

        SUM(population) OVER () AS total_population,

        SUM(population) OVER (
            ORDER BY population
            ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
        ) AS cumulative_population

    FROM valid
)

SELECT
    pixel_rank * 1.0 / total_pixels AS cumulative_pixel_share,

    cumulative_population / total_population
        AS cumulative_population_share

FROM ranked

ORDER BY pixel_rank;
""").df()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 8))

# Lorenz curve
plt.plot(
    lorenz_df["cumulative_pixel_share"],
    lorenz_df["cumulative_population_share"],
    label="Lorenz curve"
)

# Perfect equality line
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Perfect equality"
)

plt.xlabel("Cumulative share of pixels")
plt.ylabel("Cumulative share of population")
plt.title("Lorenz Curve of WorldPop Population")
plt.legend()
plt.grid(alpha=0.2)

plt.show()
```

```python
import numpy as np

x = lorenz_df["cumulative_pixel_share"].to_numpy()
y = lorenz_df["cumulative_population_share"].to_numpy()

area = np.trapezoid(y, x)

gini = 1 - 2 * area

print("Gini:", gini)
```
